Feature Engineering


```python
cleaned_df["SalesLineTotal"] = cleaned_df["Quantity"] * cleaned_df["Price"]

cleaned_df
```

```python
aggregated_df = cleaned_df.groupby(by = "Customer ID", as_index=False).agg(
    MonetaryValue=("SalesLineTotal","sum") ,
    Frequency=("Invoice","nunique"),
    LastInvoiceDate=("InvoiceDate","max")
)
aggregated_df
```

```python
max_invoice_date = aggregated_df["LastInvoiceDate"].max()

max_invoice_date
```

```python
aggregated_df["Recency"] =(max_invoice_date - aggregated_df['LastInvoiceDate']).dt.days

aggregated_df.head()
```

```python
M_Q1 = aggregated_df["MonetaryValue"].quantile(0.25)
M_Q3 = aggregated_df["MonetaryValue"].quantile(0.75)
M_IQR = M_Q3 - M_Q1

monetary_outliners_df = aggregated_df[(aggregated_df["MonetaryValue"] > (M_Q3 + 1.5 * M_IQR)) | (aggregated_df["MonetaryValue"] < (M_Q1 - 1.5 * M_IQR)) ].copy()

monetary_outliners_df.describe()
```

```python
F_Q1 = aggregated_df["Frequency"].quantile(0.25)
F_Q3 = aggregated_df["Frequency"].quantile(0.75)
F_IQR = F_Q3 - F_Q1

frequency_outliners_df = aggregated_df[(aggregated_df["Frequency"] > (F_Q3 + 1.5 * F_IQR)) | (aggregated_df["Frequency"] < (F_Q1 - 1.5 * F_IQR)) ].copy()

frequency_outliners_df.describe()
```

```python
non_outliners_df=aggregated_df[(~aggregated_df.index.isin(monetary_outliners_df.index)) & (~aggregated_df.index.isin(frequency_outliners_df.index))]
non_outliners_df.describe()
```

K-Means Clustering

```python
max_k =12

inertia =[]
silhouette_scores =[]
k_values = range(2 , max_k + 1)

for k in k_values:
    kmeans = KMeans(n_clusters= k, random_state=42, max_iter=1000)

    cluster_labels = kmeans.fit_predict(scaled_data_df)

    sil_score = silhouette_score(scaled_data_df, cluster_labels)

    silhouette_scores.append(sil_score)
    inertia.append(kmeans.inertia_)

plt.Figure(figsize=(14,20))

plt.subplot(1,2,1)
plt.plot(k_values , inertia, marker='o')
plt.title("KMeans Inertia for different values of K")
plt.xlabel('Number of Clusters(K)')
plt.ylabel('Inertia')
plt.xticks(k_values)
plt.grid(True)

plt.subplot(1,2,2)
plt.plot(k_values , silhouette_scores, marker='o', color = 'orange')
plt.title("Silhoutte Scores for different values of K")
plt.xlabel('Number of Clusters(K)')
plt.ylabel('Silhoutte Score')
plt.xticks(k_values)
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
kmeans = KMeans(n_clusters=4, random_state=42, max_iter=1000)
cluster_labels = kmeans.fit_predict(scaled_data_df)
cluster_labels
```

```python
non_outliners_df["Cluster"]= cluster_labels
non_outliners_df
```

```python
cluster_colors = {0:'#1f77b4',
                  1:'#ff7f0e',
                  2:'#2ca02c',
                  3:'#d62728'}

colors = non_outliners_df["Cluster"].map(cluster_colors)

fig = plt.figure(figsize=(10,10))
ax = fig.add_subplot(projection='3d')

scatter = ax.scatter(non_outliners_df['MonetaryValue'], 
                     non_outliners_df['Frequency'], 
                     non_outliners_df['Recency'],
                     c = colors,
                     marker = 'o')

ax.set_xlabel('Monetary Value')
ax.set_ylabel('frequency')
ax.set_zlabel('Recency')

ax.set_title("3D Scatter Plot of Customer Data by Cluster")

plt.show()
```

```python
plt.figure(figsize=(12,18))

plt.subplot(3,1,1)
sns.violinplot(x=non_outliners_df['Cluster'],y=non_outliners_df['MonetaryValue'],palette = cluster_colors , hue = non_outliners_df["Cluster"] )
sns.violinplot(y=non_outliners_df["MonetaryValue"], color='gray', linewidth = 1.0)
plt.title("Monetary Value by Cluster")
plt.ylabel("Monetary Value")

plt.subplot(3,1,2)
sns.violinplot(x=non_outliners_df['Cluster'],y=non_outliners_df['Frequency'],palette = cluster_colors , hue = non_outliners_df["Cluster"] )
sns.violinplot(y=non_outliners_df["Frequency"],color='gray', linewidth = 1.0)
plt.title("Frequency by Cluster")
plt.ylabel("Frequency")

plt.subplot(3,1,3)
sns.violinplot(x=non_outliners_df['Cluster'],y=non_outliners_df['Recency'],palette = cluster_colors , hue = non_outliners_df["Cluster"] )
sns.violinplot(y=non_outliners_df["Recency"], color='gray', linewidth = 1.0)
plt.title("Receny by Cluster")
plt.ylabel("Recency")

plt.tight_layout()
plt.show()
```

Outliner Analysis

```python
overlap_indices = monetary_outliners_df.index.intersection(frequency_outliners_df.index)

overlap_indices
```

```python
monetary_only_outliners = monetary_outliners_df.drop(overlap_indices)
frequency_only_outliners = frequency_outliners_df.drop(overlap_indices)
monetary_and_frequency_outliners = monetary_outliners_df.loc[overlap_indices]

monetary_only_outliners["Cluster"] = -1
frequency_only_outliners["Cluster"] = -2
monetary_and_frequency_outliners["Cluster"] = -3

outliner_clusters_df = pd.concat([monetary_only_outliners,
                                  frequency_only_outliners,
                                  monetary_and_frequency_outliners])

outliner_clusters_df
```

```python
cluster_colors = {-1: '#9467bd',
                  -2: '#8c564b',
                  -3: '#e377c2'}

plt.figure(figsize=(12, 18))

plt.subplot(3, 1, 1)
sns.violinplot(x=outliner_clusters_df['Cluster'], y=outliner_clusters_df['MonetaryValue'], palette=cluster_colors, hue=outliner_clusters_df["Cluster"], legend=False)
sns.violinplot(y=outliner_clusters_df["MonetaryValue"], color='gray', linewidth=1.0)
plt.title("Monetary Value by Cluster")
plt.ylabel("Monetary Value")

plt.subplot(3, 1, 2)
sns.violinplot(x=outliner_clusters_df['Cluster'], y=outliner_clusters_df['Frequency'], palette=cluster_colors, hue=outliner_clusters_df["Cluster"], legend=False)
sns.violinplot(y=outliner_clusters_df["Frequency"], color='gray', linewidth=1.0)
plt.title("Frequency by Cluster")
plt.ylabel("Frequency")

plt.subplot(3, 1, 3)
sns.violinplot(x=outliner_clusters_df['Cluster'], y=outliner_clusters_df['Recency'], palette=cluster_colors, hue=outliner_clusters_df["Cluster"], legend=False)
sns.violinplot(y=outliner_clusters_df["Recency"], color='gray', linewidth=1.0)
plt.title("Recency by Cluster")
plt.ylabel("Recency")

plt.tight_layout()
plt.show()
```

DATA VISUALISATION

```python
cluster_labels = {
    0 : 'RETAIN',
    1 : 'RE-ENGAGE',
    2 : 'NURTURE',
    3 : 'REWARD',
    -1 : 'PAMPER',
    -2 : 'UPSELL',
    -3 : 'DELIGHT'
}
```

```python
full_clustering_df = pd.concat([non_outliners_df,outliner_clusters_df])
full_clustering_df
```

```python
cluster_labels = {
    0: 'RETAIN',
    1: 'RE-ENGAGE',
    2: 'NURTURE',
    3: 'REWARD',
    -1: 'PAMPER',
    -2: 'UPSELL',
    -3: 'DELIGHT'
}

full_clustering_df["ClusterLabel"] = full_clustering_df["Cluster"].map(cluster_labels)
full_clustering_df
```

```python
cluster_counts = full_clustering_df["ClusterLabel"].value_counts()
full_clustering_df["MonetaryValue per 100 pounds"] = full_clustering_df["MonetaryValue"] / 100.00

feature_means = (full_clustering_df
                 .groupby("ClusterLabel")[["Recency", "Frequency", "MonetaryValue per 100 pounds"]]
                 .mean()
                 .reindex(cluster_counts.index))   # same order as the bars

fig, ax1 = plt.subplots(figsize=(12, 8))

sns.barplot(x=cluster_counts.index, y=cluster_counts.values, ax=ax1,
            palette='viridis', hue=cluster_counts.index, legend=False)
ax1.set_ylabel("Number of Customers", color='b')
ax1.set_title("Cluster Distribution with Average Feature Values")

ax2 = ax1.twinx()

sns.lineplot(data=feature_means, ax=ax2, palette='Set2', marker='o')
ax2.set_ylabel("Average Value", color='g')

plt.show()
```
